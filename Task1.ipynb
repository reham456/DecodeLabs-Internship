```python
# ══════════════════════════════════════════════════════════
#  TASK 1 — DATA COLLECTION & DATASET UNDERSTANDING
# ══════════════════════════════════════════════════════════
print("\n" + "═"*60)
print("  TASK 1 — DATA COLLECTION & DATASET UNDERSTANDING")
print("═"*60)

df_raw = pd.read_csv("Titanic-Dataset.csv")

print(f"\n Dataset loaded: {df_raw.shape[0]} rows × {df_raw.shape[1]} columns")
```

```python
print("\n── Column Overview ──────────────────────────────────────")
info = pd.DataFrame({
    "Column":    df_raw.columns,
    "Data Type": df_raw.dtypes.values,
    "Non-Null":  df_raw.notnull().sum().values,
    "Nulls":     df_raw.isnull().sum().values,
    "Unique":    df_raw.nunique().values,
})
print(info.to_string(index=False))
```

```python
print("\n── What Each Column Represents ──────────────────────────")
descriptions = {
    "PassengerId": "Unique identifier for each passenger",
    "Survived":    "Survival (0 = No, 1 = Yes) — TARGET variable",
    "Pclass":      "Ticket class: 1st, 2nd, or 3rd (proxy for wealth)",
    "Name":        "Full passenger name (includes title)",
    "Sex":         "Passenger gender",
    "Age":         "Age in years (177 missing values)",
    "SibSp":       "# of siblings / spouses aboard",
    "Parch":       "# of parents / children aboard",
    "Ticket":      "Ticket number (alphanumeric, messy)",
    "Fare":        "Passenger fare paid",
    "Cabin":       "Cabin number (687 missing — ~77% absent)",
    "Embarked":    "Port of embarkation: C=Cherbourg, Q=Queenstown, S=Southampton",
}
for col, desc in descriptions.items():
    print(f"  {col:<14} → {desc}")
```
